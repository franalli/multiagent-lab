```python
import itertools

text = "Ｕｎｉｃｏｄｅ! 🅤🅝🅘🅒🅞🅓🅔‽ 🇺‌🇳‌🇮‌🇨‌🇴‌🇩‌🇪! 😄 The very name strikes fear and awe into the hearts of programmers worldwide. We all know we ought to “support Unicode” in our software (whatever that means—like using wchar_t for all the strings, right?). But Unicode can be abstruse, and diving into the thousand-page Unicode Standard plus its dozens of supplementary annexes, reports, and notes can be more than a little intimidating. I don’t blame programmers for still finding the whole thing mysterious, even 30 years after Unicode’s inception."
raw_tokens = text.encode("utf-8")  # raw bytes
tokens = list(map(int, raw_tokens))  # convert to list of integers

print(text)
print(raw_tokens)
print("length:", len(text))
print("----")
print(tokens)
print("length:", len(tokens))
```

```python
def get_stats(ids):
    counts = {}
    for pair in itertools.pairwise(ids):
        counts[pair] = counts.get(pair, 0) + 1
    return counts


stats = get_stats(tokens)
print(sorted(((v, k) for k, v in stats.items()), reverse=True))
```

```python
def merge(ids, pair, idx):
    newids = []
    i = 0
    while i < len(ids):
        # if we are not at the last position and the pair matches, replace it
        if i < len(ids) - 1 and ids[i] == pair[0] and ids[i + 1] == pair[1]:
            newids.append(idx)
            i += 2
        else:
            newids.append(ids[i])
            i += 1
    return newids


# print(merge([5, 6, 6, 7, 9, 1], (6,  7), 99))

tokens2 = merge(tokens, top_pair, 256)
print(tokens2)
print("length:", len(tokens2))
```

```python
vocab_size = 276
num_merges = vocab_size - 256
ids = list(tokens)  # copy the original list to new element

merges = {}  # (int, int) -> int
for i in range(num_merges):
    stats = get_stats(ids)
    pair = max(stats, key=stats.get)
    idx = 256 + i
    print(f"merging {pair} into a new token {idx}")
    ids = merge(ids, pair, idx)
    merges[pair] = idx
```

```python
# decode
vocab = {idx: bytes([idx]) for idx in range(256)}

for (p0, p1), idx in merges.items():  # the iteration in dict is in-order
    vocab[idx] = vocab[p0] + vocab[p1]


def decode(ids):
    token_bytes = b"".join(vocab[idx] for idx in ids)
    return token_bytes.decode("utf-8", errors="replace")


print(decode(ids))
```

```python
# encode
def encode(text):
    tokens = list(text.encode("utf-8"))
    while len(tokens) >= 2:
        stats = get_stats(tokens)
        pair = min(stats, key=lambda p: merges.get(p, float("inf")))  # consecutive pairs with minimum merges
        if pair not in merges:
            break  # nothing else can be merged
        idx = merges[pair]
        tokens = merge(tokens, pair, idx)
        return tokens


print(encode("hello world"))
```

```python
# =============================================================================
# The full GPT-4 tokenizer (the cl100k_base algorithm), from scratch
# =============================================================================
# Everything above is "naive" BPE: it merges over one flat byte stream, so a
# merge is free to glue a word to whatever sits next to it -- "dog", "dog.",
# "dog!" and " dog" all end up as separate tokens and the merge table burns
# slots on punctuation variants. GPT-4 fixes this with three layers, which is
# exactly what this cell implements end to end:
#
#   1. PRE-SPLIT the text with a regex into chunks (words, numbers, punctuation
#      runs, whitespace runs). BPE then runs *inside* each chunk only, so no
#      merge can ever cross a chunk boundary.
#   2. BYTE-LEVEL BPE inside each chunk -- the same get_stats/merge loop as
#      above, just applied per chunk with one shared pair tally.
#   3. SPECIAL TOKENS (<|endoftext|> & friends) spliced in around the BPE.
#      They are never learned by BPE, so ordinary text can never forge one.

import regex as re  # NOT stdlib `re`: we need \p{L}/\p{N} + possessive quantifiers

# GPT-4's pre-tokenization pattern, verbatim from tiktoken's cl100k_base.
# Read it as alternatives tried left to right, one chunk per match:
#   '(?i:[sdmt]|ll|ve|re)      -> English contractions: 's 't 'll 've 're (any case)
#   [^\r\n\p{L}\p{N}]?+\p{L}+  -> a run of letters, allowed ONE leading non-letter,
#                                 non-digit, non-newline char (that's the space in
#                                 " hello", which is why GPT tokens start with a space)
#   \p{N}{1,3}                 -> digits in groups of at most 3, so numbers cannot
#                                 grow into arbitrary tokens ("2024" -> "202"+"4")
#   ?[^\s\p{L}\p{N}]++[\r\n]*  -> a run of punctuation/symbols, optional leading
#                                 space, trailing newlines glued on
#   \s*[\r\n]                  -> whitespace ending in a newline
#   \s+(?!\S)                  -> a whitespace run NOT followed by a non-space, i.e.
#                                 all but the last space (that last one is left to
#                                 start the next word chunk)
#   \s+                        -> any leftover whitespace (end of string case)
GPT4_SPLIT_PATTERN = r"""'(?i:[sdmt]|ll|ve|re)|[^\r\n\p{L}\p{N}]?+\p{L}+|\p{N}{1,3}| ?[^\s\p{L}\p{N}]++[\r\n]*|\s*[\r\n]|\s+(?!\S)|\s+"""


class GPT4Tokenizer:
    """Byte-level BPE with GPT-4 regex pre-splitting and special tokens."""

    def __init__(self, pattern=GPT4_SPLIT_PATTERN):
        self.compiled_pattern = re.compile(pattern)
        self.merges = {}  # (int, int) -> new id
        self.vocab = {idx: bytes([idx]) for idx in range(256)}  # id -> bytes it stands for
        self.special_tokens = {}  # "<|endoftext|>" -> id
        self.inverse_special_tokens = {}  # id -> "<|endoftext|>"

    # ---- helpers (same two primitives as above, with small tweaks) -----------

    @staticmethod
    def get_stats(ids, counts=None):
        """Count adjacent pairs. `counts` lets us tally across many chunks at once."""
        counts = {} if counts is None else counts
        for pair in itertools.pairwise(ids):
            counts[pair] = counts.get(pair, 0) + 1
        return counts

    @staticmethod
    def merge(ids, pair, idx):
        """Replace every occurrence of `pair` in `ids` with the single token `idx`."""
        newids = []
        i = 0
        while i < len(ids):
            if i < len(ids) - 1 and ids[i] == pair[0] and ids[i + 1] == pair[1]:
                newids.append(idx)
                i += 2  # consumed two tokens
            else:
                newids.append(ids[i])
                i += 1
        return newids

    # ---- training -----------------------------------------------------------

    def train(self, text, vocab_size, verbose=False):
        """Learn `vocab_size - 256` merges from `text`."""
        assert vocab_size >= 256, "the 256 raw bytes are the base vocabulary"
        num_merges = vocab_size - 256

        # Step 1: pre-split into chunks so merges stay inside word boundaries.
        text_chunks = re.findall(self.compiled_pattern, text)

        # Step 2: every chunk becomes its own list of utf-8 byte values (0..255).
        # Keeping them separate is what enforces the boundary: we never look at a
        # pair that straddles two chunks.
        ids = [list(chunk.encode("utf-8")) for chunk in text_chunks]

        merges = {}
        vocab = {idx: bytes([idx]) for idx in range(256)}

        for i in range(num_merges):
            # Step 3: tally pairs across ALL chunks into one shared dict.
            stats = {}
            for chunk_ids in ids:
                self.get_stats(chunk_ids, stats)
            if not stats:
                break  # nothing left to merge (text fully collapsed)

            # Step 4: the most frequent pair wins and gets the next free id.
            pair = max(stats, key=stats.get)
            idx = 256 + i

            # Step 5: rewrite every chunk with that pair replaced, and record it.
            ids = [self.merge(chunk_ids, pair, idx) for chunk_ids in ids]
            merges[pair] = idx
            vocab[idx] = vocab[pair[0]] + vocab[pair[1]]  # built from children, in order

            if verbose:
                print(f"merge {i + 1:>3}/{num_merges}: {pair} -> {idx} ({vocab[idx]!r}) had {stats[pair]} occurrences")

        self.merges = merges  # used by encode
        self.vocab = vocab  # used by decode

    # ---- special tokens -----------------------------------------------------

    def register_special_tokens(self, special_tokens):
        """special_tokens: dict like {"<|endoftext|>": 100257}. Ids live above the
        learned merges, are never produced by BPE, and are handled by encode/decode."""
        self.special_tokens = special_tokens
        self.inverse_special_tokens = {v: k for k, v in special_tokens.items()}

    # ---- decoding: ids -> text ----------------------------------------------

    def decode(self, ids):
        part_bytes = []
        for idx in ids:
            if idx in self.vocab:  # ordinary token: look up its bytes
                part_bytes.append(self.vocab[idx])
            elif idx in self.inverse_special_tokens:  # special token: its literal text
                part_bytes.append(self.inverse_special_tokens[idx].encode("utf-8"))
            else:
                raise ValueError(f"invalid token id: {idx}")
        text_bytes = b"".join(part_bytes)
        # errors="replace" because a slice of ids can cut a multi-byte character
        # in half -- the bytes are only guaranteed valid utf-8 for a full sequence.
        return text_bytes.decode("utf-8", errors="replace")

    # ---- encoding: text -> ids ----------------------------------------------

    def _encode_chunk(self, text_bytes):
        """BPE a single chunk's bytes, replaying merges in the order they were learned."""
        ids = list(text_bytes)
        while len(ids) >= 2:
            stats = self.get_stats(ids)
            # Pick the pair with the LOWEST merge id = the earliest merge learned.
            # Order matters: later merges are built out of earlier ones, so replaying
            # them out of order would produce tokens that training never saw.
            # Pairs we never learned map to inf, so they sort last.
            pair = min(stats, key=lambda p: self.merges.get(p, float("inf")))
            if pair not in self.merges:
                break  # even the best candidate is unknown -> nothing left to merge
            ids = self.merge(ids, pair, self.merges[pair])
        return ids

    def encode_ordinary(self, text):
        """Encode text, treating special-token strings as ordinary text."""
        ids = []
        for chunk in re.findall(self.compiled_pattern, text):  # same split as training
            ids.extend(self._encode_chunk(chunk.encode("utf-8")))
        return ids

    def encode(self, text, allowed_special="all"):
        """Encode text, honouring special tokens. allowed_special: "all" or "none"."""
        special = self.special_tokens if allowed_special == "all" else {}
        if not special:
            return self.encode_ordinary(text)

        # Split on the special strings, KEEPING them (the capture group does that),
        # so we get e.g. ["hello ", "<|endoftext|>", " world"].
        special_pattern = "(" + "|".join(re.escape(k) for k in special) + ")"
        ids = []
        for part in re.split(special_pattern, text):
            if part in special:
                ids.append(special[part])  # exact id, no BPE
            elif part:
                ids.extend(self.encode_ordinary(part))  # normal BPE path
        return ids


# ---- demo: train on the same text as the naive run above ---------------------

tokenizer = GPT4Tokenizer()
tokenizer.train(text, vocab_size=276)  # same budget: 20 merges
tokenizer.register_special_tokens({"<|endoftext|>": 100257})  # GPT-4's real id for it

print("how the regex chops text:")
print(re.findall(GPT4_SPLIT_PATTERN, "Hello world123!!! It's 2024  "))
print("\nthe 20 new tokens it learned:")
print([tokenizer.vocab[idx] for idx in range(256, 276)])

# Compression is slightly WORSE than the 1.37X of the naive run, and that is the
# point: the naive winner was (101, 32) = "e" + space, a merge that welds a word
# to the space after it. The regex forbids those, so on 533 characters we lose a
# little. At GPT-4 scale the trade pays for itself many times over -- the vocab
# stops spending slots on "dog", "dog.", "dog!", " dog" as four unrelated tokens.
full_ids = tokenizer.encode_ordinary(text)
n_bytes = len(text.encode("utf-8"))
print(f"\nbytes: {n_bytes} -> tokens: {len(full_ids)} (compression {n_bytes / len(full_ids):.2f}X)")

# Roundtrip, with a special token sitting in the middle of ordinary text.
demo = "hello world!!!? (안녕하세요!) 😉 <|endoftext|> the end"
demo_ids = tokenizer.encode(demo)
print("\nids:   ", demo_ids)
print("pieces:", [tokenizer.decode([i]) for i in demo_ids])  # '�' = token holds a
#                                                              partial utf-8 char
print("roundtrip ok:", tokenizer.decode(demo_ids) == demo)

# What separates this from the real GPT-4 tokenizer is only scale and provenance:
# identical algorithm, but vocab_size=100277, merges trained on internet-scale
# data (importable from `tiktoken`, not re-derivable here), plus one historical
# quirk -- cl100k_base's 256 single-byte tokens are shuffled by a fixed permutation.
```

```python
# =============================================================================
# The same job the other way: SentencePiece -- how Mistral/Llama train theirs
# =============================================================================
# GPT-4 above:   text -> UTF-8 BYTES  -> BPE over bytes           (tiktoken)
# Mistral here:  text -> UNICODE CODE POINTS -> BPE over code points, with rare
#                code points falling back to bytes                (sentencepiece)
#
# Same BPE core. The difference is what the *atoms* are, plus a pile of extra
# machinery sentencepiece wraps around them -- normalization, a dummy space
# prefix, its own whitespace marker, character coverage, an <unk> token. Many of
# those options are archaeology from the pre-LLM machine-translation era. The
# distinctive settings below -- identity normalization, byte_fallback, character
# coverage, split_digits, dummy prefix -- are the ones you can read back out of
# the Llama 2 / Mistral v1 tokenizer.model protobuf, and they mostly amount to
# "turn the archaeology off". The rest are sentencepiece defaults, spelled out
# here so the whole configuration is visible in one place.

import os

import sentencepiece as spm

# Step 1: sentencepiece trains from a FILE on disk (one "sentence" per line),
# not from a Python string -- so dump the same text we used above.
with open("toy.txt", "w", encoding="utf-8") as f:
    f.write(text)

options = {
    # --- input / output -----------------------------------------------------
    "input": "toy.txt",
    "input_format": "text",
    "model_prefix": "tok400",  # writes tok400.model (binary) + tok400.vocab (text)
    # --- the algorithm ------------------------------------------------------
    "model_type": "bpe",  # the same byte-pair merging as above.
    # (sentencepiece's DEFAULT is "unigram", a
    #  different algorithm entirely -- it starts
    #  from a big vocab and prunes it down.)
    "vocab_size": 400,  # Mistral v1 / Llama 2 shipped 32000. The floor
    # here is 3 specials + 256 byte pieces + every
    # covered character, so asking for 300 on this
    # text fails: "Vocabulary size is smaller than
    # required_chars. 300 vs 320".
    # --- normalization: Llama/Mistral turn it off ---------------------------
    "normalization_rule_name": "identity",  # do NOT NFKC-fold the text. Default
    # would rewrite quotes, fullwidth chars,
    # etc. -- lossy before you even tokenize.
    "remove_extra_whitespaces": False,  # keep runs of spaces exactly as written
    # --- rare-character handling -- the important pair ----------------------
    "character_coverage": 0.99995,  # keep the code points covering 99.995% of the
    # training text as atoms; the long tail is
    # dropped from the vocab...
    "byte_fallback": True,  # ...and instead of becoming <unk>, a dropped
    # character is re-encoded as its raw UTF-8
    # bytes, each a piece like <0xEC>. This is what
    # makes the tokenizer lossless despite being
    # code-point based. 256 vocab slots pay for it.
    # --- what BPE is allowed to merge across --------------------------------
    "split_by_whitespace": True,  # merges never cross a space...
    "split_by_unicode_script": True,  # ...nor a script change (Latin/Hangul/emoji)
    "split_by_number": True,  # ...nor a letter/digit boundary
    "split_digits": True,  # every digit is its own piece: "2024" ->
    # 2,0,2,4. GPT-4's regex instead caps runs at 3.
    "max_sentencepiece_length": 16,  # no token longer than 16 characters
    "add_dummy_prefix": True,  # prepend a space to every input, so "hello"
    # and " hello" tokenize identically
    "allow_whitespace_only_pieces": True,  # let runs of spaces become one token
    # --- training-set sampling (irrelevant here, matters on a real corpus) ---
    "input_sentence_size": 200000000,
    "shuffle_input_sentence": True,
    "seed_sentencepiece_size": 1000000,
    "max_sentence_length": 4192,
    # --- special tokens: these IDs are Mistral's/Llama's exactly ------------
    "unk_id": 0,  # <unk>  -- must exist, even when byte_fallback makes it unreachable
    "bos_id": 1,  # <s>    -- beginning of sequence
    "eos_id": 2,  # </s>   -- end of sequence
    "pad_id": -1,  # no pad token (-1 turns one off)
    "num_threads": os.cpu_count(),
    # minloglevel=1,             # uncomment to silence the ~300-line training log
}

spm.SentencePieceTrainer.train(**options)  # prints its training log below

# Step 2: load the trained model and look at how the vocabulary is laid out.
sp = spm.SentencePieceProcessor()
sp.load("tok400.model")

print("\nvocab size:", sp.get_piece_size())
print("ids   0-2 (specials):", [sp.id_to_piece(i) for i in range(3)])
print("ids   3-8 (byte fallback, 256 of them, 3..258):", [sp.id_to_piece(i) for i in range(3, 9)])
print("ids 259+  (learned merges):", [sp.id_to_piece(i) for i in range(259, 275)])
# The '▁' in those pieces is U+2581 LOWER ONE EIGHTH BLOCK, NOT a space: before
# training, sentencepiece rewrites every space as '▁' so whitespace survives as
# an ordinary character and word boundaries stay visible inside tokens.

# Step 3: encode/decode the same demo string the GPT-4 tokenizer handled above.
demo = "hello world!!!? (안녕하세요!) 😉 <|endoftext|> the end"
spm_ids = sp.encode(demo, add_bos=True, add_eos=True)  # <s> ... </s>

print("\nsentencepiece ids:   ", spm_ids)
print("sentencepiece pieces:", [sp.id_to_piece(i) for i in spm_ids])
print("roundtrip ok:", sp.decode(spm_ids) == demo)
# Note what happened to the Hangul: it never appeared in our 533-character
# training text, so it was not covered -- each character comes back as three
# <0x..> byte pieces. That is byte_fallback earning its 256 slots. With
# byte_fallback=False every one of them would be a single lossy <unk>.
# Note also '<|endoftext|>': sentencepiece has no notion of special strings at
# encode time, so it tokenizes the literal characters. To make it atomic you
# would pass user_defined_symbols=["<|endoftext|>"] at TRAINING time.

# Step 4: the two tokenizers, same string, side by side.
print("\ngpt-4 style :", [tokenizer.decode([i]) for i in tokenizer.encode_ordinary(demo)])
print("mistral     :", [sp.id_to_piece(i) for i in sp.encode(demo)])

# -----------------------------------------------------------------------------
# GPT-4 / tiktoken                      | Mistral v1 / sentencepiece
# --------------------------------------|--------------------------------------
# atoms are the 256 bytes, always       | atoms are code points; bytes only as
#                                       | fallback for uncovered ones
# <unk> cannot exist                    | <unk> exists (dead when byte_fallback)
# space glued on by the split regex     | space rewritten to '▁' before training
# raw text in, raw text out             | text normalized first (here: identity)
# specials live outside BPE, in code    | specials are real rows in the vocab
# digits in runs of <=3                 | digits always split individually
#
# Footnote: Mistral moved on. v1/v2 use this sentencepiece model (32k vocab);
# from Mistral NeMo onward their "Tekken" tokenizer is tiktoken-style byte-level
# BPE with a 131k vocab -- i.e. the cell above, not this one.
```
