```python
with open("input.txt", "r", encoding="utf-8") as f:
    text = f.read()
```

```python
print(text[:1000])
```

```python
# vocabulary
chars = sorted(set(text))
vocab_size = len(chars)
print("".join(chars))
print("vocab size:", vocab_size)
```

```python
# character-level encoding and decoding
stoi = {ch: i for i, ch in enumerate(chars)}
itos = {i: ch for i, ch in enumerate(chars)}

encode = lambda s: [stoi[c] for c in s]
decode = lambda l: "".join([itos[i] for i in l])

print(encode("hello"))
print(decode(encode("hello")))
```

```python
import torch

data = torch.tensor(encode(text), dtype=torch.long)
print(data[:1000])
```

```python
n = int(0.9 * len(data))
train_data = data[:n]
val_data = data[n:]
print("train data length:", len(train_data))
print("validation data length:", len(val_data))
```

```python
block_size = 8
train_data[: block_size + 1]
```

```python
x = train_data[:block_size]  # the training input sequence
y = train_data[1 : block_size + 1]  # targets, offset by one
for t in range(block_size):
    context = x[: t + 1]
    target = y[t]
    print(f"when input is {context} the target: {target}")
```

```python
torch.manual_seed(1337)
batch_size = 4  # independent sequences in parallel
block_size = 8  # length of each context sequence


def get_batch(split):
    data = train_data if split == "train" else val_data
    ix = torch.randint(len(data) - block_size, (batch_size,))
    x = torch.stack([data[i : i + block_size] for i in ix])
    y = torch.stack([data[i + 1 : i + block_size + 1] for i in ix])
    return x, y


xb, yb = get_batch("train")
print("inputs:")
print(xb.shape)
print(xb)
print("targets:")
print(yb.shape)
print(yb)
print("-----")

for b in range(batch_size):  # batch dimension
    for t in range(block_size):  # time dimension
        context = xb[b, : t + 1]
        target = yb[b, t]
        print(f"when input is {context.tolist()} the target: {target}")
```

```python
import torch
from torch import nn
from torch.nn import functional as F

torch.manual_seed(1337)


class BigramLanguageModel(nn.Module):
    # This model has no history; it predicts the next token based solely on the current token
    def __init__(self, vocab_size):
        super().__init__()
        # the row i of the embedding table corresponds to the logits for the i-th token in the vocabulary
        self.token_embedding_table = nn.Embedding(vocab_size, vocab_size)

    def forward(self, idx, targets=None):
        logits = self.token_embedding_table(idx)  # (B=4, T=8, C=65)
        if targets is None:
            loss = None
        else:
            B, T, C = logits.shape
            logits = logits.view(B * T, C)
            targets = targets.view(B * T)  # target is the next token in the sequence
            loss = F.cross_entropy(logits, targets)

        return logits, loss

    def generate(self, idx, max_new_tokens):
        for _ in range(max_new_tokens):
            # get the predictions
            logits, _loss = self(idx)  # calling forward pass of the model

            # focus on the last time step because the model predicts the next token based solely on the current token
            # we ignore the full token history of the context size and only consider the last token
            logits = logits[:, -1, :]  # becomes (B, C)

            # apply softmax to get probabilities on the channel dimension
            probs = F.softmax(logits, dim=-1)  # (B, C)

            # sample from the distribution
            idx_next = torch.multinomial(probs, num_samples=1)  # (B, 1)

            # append sampled index to the sequence
            idx = torch.cat((idx, idx_next), dim=1)  # (B, T+1)
        return idx


m = BigramLanguageModel(vocab_size)
logits, loss = m(xb, yb)
print(logits.shape)
print(loss)

print(
    decode(
        m.generate(idx=torch.zeros((1, 1), dtype=torch.long), max_new_tokens=100)[
            0
        ].tolist()
    )
)
```

```python
optimizer = torch.optim.AdamW(m.parameters(), lr=1e-3)
```

```python
batch_size = 32
for steps in range(10000):
    # sample data
    xb, yb = get_batch("train")

    # evaluate loss
    logits, loss = m(xb, yb)
    optimizer.zero_grad(set_to_none=True)
    loss.backward()
    optimizer.step()

print(loss.item())
```

```python
print(
    decode(
        m.generate(idx=torch.zeros((1, 1), dtype=torch.long), max_new_tokens=500)[
            0
        ].tolist()
    )
)
```

```python
### Mathematical trick in Self-attention

torch.manual_seed(1337)
B, T, C = 4, 8, 2
x = torch.randn(B, T, C)
x.shape
```

```python
# We want x[b,t] = mean_{i<=t} x[b,i] to encapsulate the information from all previous time steps up to t
xbow = torch.zeros((B, T, C))
for b in range(B):
    for t in range(T):
        xprev = x[b, : t + 1]  # (t, C)
        xbow[b, t] = torch.mean(xprev, 0)
```

```python
# Efficient Matrix version
wei = torch.tril(torch.ones(T, T))  # lower triangular matrix
wei = wei / wei.sum(1, keepdim=True)
xbow2 = wei @ x  # (B, T, T) @ (B, T, C) -> (B, T, C)
torch.allclose(xbow, xbow2)
```

```python
# Third version (most useful)

tril = torch.tril(torch.ones(T, T))
tril
```

```python
wei = torch.zeros((T, T))
wei = wei.masked_fill(
    tril == 0, float("-inf")
)  # this means future cannot communicate with the past
wei
```

```python
wei = F.softmax(wei, dim=-1)
wei
```

```python
xbow3 = wei @ x
torch.allclose(xbow, xbow3)
```

```python
# Self attention
torch.manual_seed(1337)
B, T, C = 4, 8, 32
x = torch.randn(B, T, C)

# single head
head_size = 16
key = nn.Linear(C, head_size, bias=False)
query = nn.Linear(C, head_size, bias=False)
value = nn.Linear(C, head_size, bias=False)

k = key(x)  # (B, T, C) @ (C, H) -> (B, T, H)
q = query(x)  # (B, T, C) @ (C, H) -> (B, T, H)
v = value(x)  # (B, T, C) @ (C, H) -> (B, T, H)

wei = q @ k.transpose(-2, -1) * head_size**-0.5  # (B, T, H) @ (B, H, T) -> (B, T, T)

tril = torch.tril(torch.ones(T, T))
# wei = torch.zeros((T,T)) # token affinities start as 0
wei = wei.masked_fill(
    tril == 0, float("-inf")
)  # decoder masking. If this were an encoder, we would delete this
wei = F.softmax(wei, dim=-1)

out = wei @ v  # (B, T, T) @ (B, T, H) -> (B, T, H)
out.shape
```
